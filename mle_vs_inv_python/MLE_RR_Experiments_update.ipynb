## Imports

```python
import numpy as np
from itertools import product
from tqdm import tqdm
from sklearn.metrics import mean_squared_error
from joblib import Parallel, delayed
import joblib
import multiprocessing
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")
import matplotlib.pyplot as plt
import matplotlib

params = {
    "legend.fontsize": "medium",
    "lines.linewidth": "1.5",
    "font.weight": "normal",
    "lines.markersize": "5",
    "text.latex.preamble": r"\usepackage{amsfonts}",
    "legend.handletextpad": 0.3,
    "legend.labelspacing": 0.2,
}
matplotlib.rcParams.update(params)


def set_all_font_sizes(size):
    params = {
        "axes.titlesize": size,
        "axes.labelsize": size,
        "font.size": size,
        "figure.titlesize": size,
        "xtick.labelsize": size,
        "ytick.labelsize": size,
    }
    plt.rcParams.update(params)
    return

set_all_font_sizes(24)
plt.rcParams["mathtext.fontset"] = "cm"
plt.rc("text", usetex=True)
plt.rc("font", family="serif")

# Camera-ready settings for plots
plt.rcParams.update(
    {
        "pdf.fonttype": 42,  # TrueType
        "ps.fonttype": 42,
        "font.family": "serif",
        "font.serif": ["Times"],  # or "Nimbus Roman"
        "mathtext.fontset": "stix",  # STIX is Times-compatible, vector, no Type-3
    }
)

# Local Imports
from krr import grr_client_vec
from estimators import MLE, INV_P, INV_N
from utils import log_likelihood, generate_synthetic_data
import cfg
```

## Parameters for the experiments

```python
algorithms = {
            "MLE*": MLE, # Our MLE estimator
            "InvP": INV_P, # INV with projection into the simplex
            "InvN": INV_N, # INV with normalization
            }

metrics_local = {
            "neg_log_likelihood": lambda _, est, phi, p_truth: -log_likelihood(phi, est, p_truth),
            "MSE": lambda truth, est, *_: mean_squared_error(truth, est),
            }
metrics_global = {
    "bias_squared": lambda truth, ests: np.linalg.norm(np.mean(ests, 0) - truth)**2,
    "variance": lambda truth, ests: np.linalg.norm(np.std(ests, 0))**2,
}
metric_names = {
    # "neg_log_likelihood": "Negative Log-Likelihood",
    "neg_log_likelihood": "Neg. Log-Like.",
    # "MSE": "Mean Squared Error (MSE)",
    "MSE": "MSE",
    "bias_squared": "Bias$^2$",
    "variance": "Variance",
}

color_map = {
            "MLE*": "blue",
            "InvP": "orange",
            "InvN": "green",
        }

linestyle_map = {
    "MLE*": "-",
    "InvP": "--",
    "InvN": ":",
}

hatch_map = {
    "MLE*": "///",
    "InvP": "\\\\",
    "InvN": "xx",
}

marker_map = {
    "MLE*": "s",
    "InvP": "o",
    "InvN": "d",
}

n_repeats = cfg.n_repeats # number of seeds to repeat each experiment
```

## Function to run experiments in parallel

```python
def run_experiment_ij(true_resp, true_hist, eps, N, K, i, j, alg):

    # Perturb with GRR
    rng = np.random.default_rng(seed=j)
    p_truth = np.exp(eps) / (np.exp(eps) + K - 1)
    rand_resp = grr_client_vec(true_resp, K, p_truth, rng=rng)
    phi = np.bincount(rand_resp, minlength=K) / N

    # Estimate
    f_alg = algorithms[alg]
    theta = f_alg(phi, p_truth)
    assert np.isclose(np.sum(theta), 1), (theta, alg, i, j)

    metrics_results = {}
    for met, f_met in metrics_local.items():
        metrics_results[met] = f_met(true_hist, theta, phi, p_truth)
    return {
        "metrics": metrics_results,
        "theta": theta,
    }


def run_experiment(data, i, n_repeats, alg):
    eps, N, K, s = pops[i]
    if data is None:
        # Generate true data and histogram
        true_resp = generate_synthetic_data(N, K, s, seed=i)

    else:
        # Generate true data and histogram
        true_resp = data[s]  # .sample(n=N, random_state=j, replace=True).values
        true_resp = pd.cut(true_resp, bins=K, labels=False, include_lowest=True)
        N = len(true_resp)

    true_counts = np.bincount(true_resp, minlength=K)
    true_hist = true_counts / N

    info = {
        "population": i,
        "eps": eps,
        "N": N,
        "K": K,
        "s": s,
        "algorithm": alg,
    }
    thetas = []
    results_ij = []
    for j in range(n_repeats):
        res = run_experiment_ij(true_resp, true_hist, eps, N, K, i, j, alg)
        for met, value in res["metrics"].items():
            results_ij.append({**info, "metric": met, "value": value})
        thetas.append(res["theta"])
    thetas = np.asarray(thetas)

    results_i = []
    for met, f_met in metrics_global.items():
        value = f_met(true_hist, thetas)
        results_i.append({**info, "metric": met, "value": value})
    
    true_hist = {**info, "true_hist": true_hist}

    return results_ij, results_i, true_hist
```

## Run Experiments (Synthetic data)

```python
pop_eps = cfg.pop_eps  # Epsilon values to vary
pop_N = cfg.pop_N  # Population sizes to vary
pop_K = cfg.pop_K  # Domain sizes to vary
pop_s = cfg.pop_s  # Concentration parameters (Zipf exponents) to vary

# Each combination is (eps, N, K, s)
pops = list(product(pop_eps, pop_N, pop_K, pop_s))
combinations = list(product(range(len(pops)), algorithms))
dic_true_hist = {s: {K: None for K in pop_K} for s in pop_s}

path = "results/overview.joblib"
try:
    loaded = joblib.load(path)
except:
    loaded = None

if loaded is None:
    # Run in parallel
    num_cores = multiprocessing.cpu_count()
    results = Parallel(n_jobs=num_cores)(
        delayed(run_experiment)(None, i, n_repeats, alg)
        for i, alg in tqdm(combinations)
    )
    results_i = [res for _, res_i, _ in results for res in res_i]
    results_ij = [res for res_i, _, _ in results for res in res_i]

    # Unpack results
    dic_true_hist = {s: {K: None for K in pop_K} for s in pop_s}
    for _, _, res in results:
        # Overwrite or store the final version
        dic_true_hist[res["s"]][res["K"]] = res["true_hist"]

    loaded = (results_i, results_ij, dic_true_hist)
    joblib.dump(loaded, path)
```

## Plotting

### Overview of results: fixed_eps

```python
import matplotlib.ticker as mticker

results_i, results_ij, dic_true_hist = joblib.load("results/overview.joblib")
set_all_font_sizes(24)

# Set fixed parameters
K_fixed = 10000
eps_fixed = 4.0
s_list = [0.01, 1.3, 2.5]  # uniform, moderate, spike

for met, out in [*product(metrics_local, [results_ij]), *product(metrics_global, [results_i])]:

    # Create a figure with one row and three columns (one subplot per s)
    fig, axes = plt.subplots(
        1, len(s_list),
        figsize=(15, 4.9),
    )

    # get results into a DataFrame
    df_all = pd.DataFrame(out)

    for col_idx, s in enumerate(s_list):
        ax = axes[col_idx]
        plt.sca(ax)

        df = df_all.query(
            "metric == @met and eps == @eps_fixed and K == @K_fixed and s == @s"
        )

        if df.empty:
            ax.set_title(f"No data: eps={eps_fixed}, K={K_fixed}, s={s}")
            continue

        for alg in df.algorithm.unique():
            
            c = color_map.get(alg)
            mkr = marker_map.get(alg)
            hatch = hatch_map.get(alg)
            ls = linestyle_map.get(alg)

            df_line = df.query("algorithm == @alg").copy()
            sns.lineplot(
                data=df_line,
                x="N",
                y="value",
                color=c,
                linestyle=ls,
                estimator="mean",
                errorbar=("pi", 95),
                err_kws={"alpha": 0.25, "hatch": hatch},
                marker=mkr,
                markersize=8,
                label=alg,
                lw=2,
            )

        if s <=0.25:
            ax.set_title(f"Low Concentration, $s = {s}$")
        elif 0.8 <= s <= 1.3:
            ax.set_title(f"Moderate, $s = {s}$")
        else:
            ax.set_title(f"High, $s = {s}$")
        if met not in ["neg_log_likelihood"]:
            ax.set_yscale("log")
        ax.set_xscale("log")
        ax.xaxis.set_major_locator(mticker.LogLocator(numticks=3))
        ax.set_ylabel(metric_names.get(met, met))
        if col_idx > 0:
            ax.set_ylabel("")
        if col_idx < len(s_list) - 1:
            # remove legend
            leg = ax.get_legend()
            if leg is not None:
                leg.remove()
        else:
            # Place it outside the plot
            ax.legend(loc='center left', bbox_to_anchor=(1, 0.5))

    plt.suptitle(fr"Estimator Comparison (K={K_fixed}, $\epsilon$={eps_fixed})", y=0.85)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig(f"results/overview_met_{met}_K{K_fixed}_eps{eps_fixed}.pdf", dpi=500, bbox_inches='tight')
    plt.show()
```

### All results

```python
results_i, results_ij, dic_true_hist = joblib.load("results/overview.joblib")
set_all_font_sizes(24)

# Define the filtered epsilon values
filtered_eps = [1, 2, 4, 10]

for met, out in [*product(metrics_local, [results_ij]), *product(metrics_global, [results_i])]:
    print(f"Metric: {met}")
    df_all = pd.DataFrame(out)
    for s in pop_s:
        fig_rows = pop_K
        fig_columns = [np.nan] + filtered_eps  # only keep desired epsilons

        fig, axes = plt.subplots(
            len(fig_rows),
            len(fig_columns),
            figsize=(5 * len(fig_columns), 4 * len(fig_rows)),
            squeeze=False,
        )

        for row_idx, K in enumerate(fig_rows):
            for col_idx, eps in enumerate(fig_columns):
                ax = axes[row_idx][col_idx]
                plt.sca(ax)

                if eps is np.nan:
                    x_vals = np.arange(K)
                    ax.plot(x_vals, dic_true_hist[s][K], label=f"s={s}")
                    ax.set_title(f"Zipf, K={K}, s={s}")
                    ax.set_ylabel("Frequency")
                    ax.set_xlabel("Value")
                    if s <= 0.25:
                        ax.set_ylim(-0.2, 0.2)
                        ax.set_title(f"~Uniform, K={K}, s={s}")
                    if s >= 5:
                        ax.set_title(f"Spike, K={K}, s={s}")
                    continue

                df = df_all.query(
                    "metric == @met and eps == @eps and K == @K and s == @s"
                )
                if df.empty:
                    ax.set_title(f"No data: $\\epsilon$={eps}, K={K}, s={s}")
                    continue

                for alg in df.algorithm.unique():
                    c = color_map.get(alg)
                    mkr = marker_map.get(alg)
                    hatch = hatch_map.get(alg)
                    ls = linestyle_map.get(alg)

                    df_line = df.query("algorithm == @alg").copy()
                    sns.lineplot(
                        data=df_line,
                        x="N",
                        y="value",
                        color=c,
                        linestyle=ls,
                        estimator="mean",
                        errorbar=("pi", 95),
                        err_kws={"alpha": 0.25, "hatch": hatch},
                        marker=mkr,
                        label=alg,
                        lw=2,
                    )

                ax.set_title(f"K={K}, $\\epsilon$={eps}, s={s}")
                if met not in ["neg_log_likelihood"]:
                    ax.set_yscale("log")
                ax.set_xscale("log")
                ax.set_ylabel(met)
                if col_idx > 0:
                    ax.set_ylabel("")
                ax.set_xlabel("N (population size)")

        plt.tight_layout()
        plt.grid(True)
        plt.savefig(f"results/zipf_dataset_s_{s}_{met}.pdf", dpi=500, bbox_inches='tight')
        plt.show()

    print("==========================================\n")
```
